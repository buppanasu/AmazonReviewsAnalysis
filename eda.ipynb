```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Step 1: Convert the timestamp to datetime ---
# Assuming the timestamp is in centiseconds; adjust if needed.
if pd.api.types.is_datetime64_any_dtype(df['timestamp']):
    df['timestamp_dt'] = df['timestamp']
else:
    df['timestamp_dt'] = pd.to_datetime(df['timestamp'] / 100, unit='s')

# --- Step 2: Group by user and minute intervals ---
# Use 'min' for minute-level grouping (the 'T' alias is deprecated)
user_minute_counts = df.groupby(['user_id', pd.Grouper(key='timestamp_dt', freq='min')]) \
                       .size() \
                       .reset_index(name='review_count')

# --- Step 3: Compute per-user statistics and z-scores ---
# Compute each user's mean and standard deviation of review counts per minute
user_stats = user_minute_counts.groupby('user_id')['review_count'] \
                               .agg(user_mean='mean', user_std='std') \
                               .reset_index()

# Merge the statistics back into the grouped data
merged = pd.merge(user_minute_counts, user_stats, on='user_id', how='left')

# Compute the z-score for each minute for each user
# Handle cases where user_std might be zero (e.g., if a user only posts 1 review per minute)
merged['z_score'] = (merged['review_count'] - merged['user_mean']) / merged['user_std'].replace(0, np.nan)

# Flag potential spam events using a z-score threshold (e.g., z > 3)
# You can adjust the threshold as needed.
merged['is_spam'] = merged['z_score'] > 3

# --- Step 4: Visualization ---
# Scatter plot: time on x-axis, review count on y-axis, color by whether the activity is flagged as spam
plt.figure(figsize=(14, 7))
sns.scatterplot(data=merged, x='timestamp_dt', y='review_count',
                hue='is_spam', palette={True: 'red', False: 'blue'}, alpha=0.6)
plt.title("User Review Activity Over Time (Per Minute) - Spam Highlighted")
plt.xlabel("Time")
plt.ylabel("Review Count (per minute)")
plt.legend(title='Suspicious', labels=['Normal', 'Potential Spam'])
plt.show()

# --- Step 5: Identify Top Spam Events ---
# Display the top few instances flagged as potential spam
top_spam_events = merged[merged['is_spam']].sort_values(by='review_count', ascending=False)
print("Top Spam Events:\n", top_spam_events.head(10))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Step 1: Ensure timestamp is in datetime format ---
if pd.api.types.is_datetime64_any_dtype(df['timestamp']):
    df['timestamp_dt'] = df['timestamp']
else:
    df['timestamp_dt'] = pd.to_datetime(df['timestamp'] / 100, unit='s')

# --- Step 2: Extract the year and group by it ---
df['year'] = df['timestamp_dt'].dt.year
yearly_counts = df.groupby('year').size().reset_index(name='review_count')

# --- Step 3: Print the reference table ---
print("Review Counts by Year:")
print(yearly_counts)

# --- Step 4: Plot the review counts by year ---
plt.figure(figsize=(10, 6))
sns.barplot(data=yearly_counts, x='year', y='review_count', palette='viridis')
plt.title("Review Counts by Year")
plt.xlabel("Year")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load JSON Lines
df = pd.read_json('/Users/junjie/lab10spark/Video_Games1.jsonl', lines=True)

# Parse timestamp and filter year
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df[df['timestamp'].dt.year == 2023]

# Extract hour of day
df['hour'] = df['timestamp'].dt.hour

# Create pivot: index=hour, columns=reviewer_id, values=count of reviews
pivot = df.pivot_table(index='hour',
                       columns='user_id',
                       values='timestamp',
                       aggfunc='count',
                       fill_value=0)

# Identify top 10 most active reviewers
top_reviewers = pivot.sum().nlargest(10).index

# Plot each reviewer’s hourly review counts
plt.figure(figsize=(12, 6))
for reviewer in top_reviewers:
    plt.plot(pivot.index, pivot[reviewer], label=reviewer)

plt.xlabel('Hour of Day')
plt.ylabel('Number of Reviews')
plt.title('Reviews Posted per Hour by Top 10 Users (2023)')
plt.xticks(range(24))
plt.legend(title='Reviewer ID', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
